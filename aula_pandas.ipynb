```python
def classificar_qntd(qntd): # Criamos uma função para classificar o tipo de venda de acordo com a quantidade
    if qntd == 1:
        return 'Varejo'
    elif qntd <= 4:
        return 'Média'
    else:
        return 'Atacado'
```

```python
df['tipo_venda'] = df['quantidade'].apply(classificar_qntd) # Aplica-se a função e coloca na coluna "tipo_venda"
```
